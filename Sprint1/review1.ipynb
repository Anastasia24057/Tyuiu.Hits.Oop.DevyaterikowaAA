Создать проект на C#, реализующий классы Товар (Product), Продавец (Seller), Магазин (Store) с использование геттеров и сеттеров, а также модификаторов доступа public, private, protected, internal. От каждого класса создайть по одному экземпляру.  

```
using System;
using System.Collections.Generic;

try
{
    Product product = new Product("Помада", 450m, 20, "Косметика", "Матовая помада");
    Seller seller = new Seller("Анна", "05667", 45000m, "anna@store.com");
    Store store = new Store("Бьюти Маркет", "ул. Ленина, д. 10", "09:00 - 21:00");

    Console.WriteLine(store.GetStoreInfo());
    Console.WriteLine(seller.GetSellerInfo());
    Console.WriteLine(product.GetProductInfo());
    Console.WriteLine($"Начальная общая стоимость товара: {product.GetTotalPrice()} руб.");
    Console.WriteLine(new string('-', 75));

    store.AddSeller(seller);
    seller.AddProduct(product);

    seller.SellProduct(product, 5);
    Console.WriteLine($"[После продажи]: {product.GetProductInfo()}");
    Console.WriteLine($"Новая общая стоимость товара: {product.GetTotalPrice()} руб.");
    Console.WriteLine(new string('-', 75));

    product.UpdateQuantity(10);
    Console.WriteLine($"[После поставки]: {product.GetProductInfo()}");
    Console.WriteLine(new string('-', 75));

    Console.WriteLine("Список всех товаров в магазине:");
    store.ListProducts();
}
catch (Exception ex)
{
    Console.WriteLine($"Ошибка: {ex.Message}");
}


public class Product
{
    private string name;
    private decimal price;
    private int quantity;
    protected string category;
    internal string description;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public decimal Price
    {
        get { return price; }
        set { price = value; }
    }

    public int Quantity
    {
        get { return quantity; }
        set { quantity = value; }
    }

    public string Category
    {
        get { return category; }
        set { category = value; }
    }

    public string Description
    {
        get { return description; }
        set { description = value; }
    }

    public Product(string name, decimal price, int quantity, string category, string description)
    {
        Name = name;
        Price = price;
        Quantity = quantity;
        Category = category;
        Description = description;
    }

    public decimal GetTotalPrice()
    {
        return Price * Quantity;
    }

    public void UpdateQuantity(int amount)
    {
        Quantity += amount;
        Console.WriteLine($"Количество товара '{Name}' изменено на {amount} шт.");
    }

    public string GetProductInfo()
    {
        return $"Товар: {Name} | Категория: {Category} | Цена: {Price} руб. | Доступно: {Quantity} шт. | Описание: {Description}";
    }
}

public class Seller
{
    private string name;
    private string employeeId;
    protected decimal salary;
    internal string contactInfo;
    
    private List<Product> productsList = new List<Product>();

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public string EmployeeId
    {
        get { return employeeId; }
        set { employeeId = value; }
    }

    public decimal Salary
    {
        get { return salary; }
        set { salary = value; }
    }

    public string ContactInfo
    {
        get { return contactInfo; }
        set { contactInfo = value; }
    }

    public List<Product> ProductsList
    {
        get { return productsList; }
    }

    public Seller(string name, string employeeId, decimal salary, string contactInfo)
    {
        Name = name;
        EmployeeId = employeeId;
        Salary = salary;
        ContactInfo = contactInfo;
    }

    public void AddProduct(Product product)
    {
        productsList.Add(product);
        Console.WriteLine($"Продавец {Name} добавил товар '{product.Name}' в свой список.");
    }

    public void SellProduct(Product product, int quantity)
    {
        if (product.Quantity >= quantity)
        {
            product.Quantity -= quantity;
            Console.WriteLine($"Продавец {Name} продал товар '{product.Name}' в количестве {quantity} шт.");
        }
        else
        {
            Console.WriteLine($"Ошибка: Недостаточно товара '{product.Name}' на складе!");
        }
    }

    public string GetSellerInfo()
    {
        return $"Продавец: {Name} (ID: {EmployeeId}) | Контакты: {ContactInfo}";
    }
}

public class Store
{
    private string storeName;
    private string location;
    public string storeHours;
    
    private List<Seller> sellersList = new List<Seller>();

    public string StoreName
    {
        get { return storeName; }
        set { storeName = value; }
    }

    public string Location
    {
        get { return location; }
        set { location = value; }
    }

    public Store(string storeName, string location, string storeHours)
    {
        StoreName = storeName;
        Location = location;
        this.storeHours = storeHours;
    }

    public void AddSeller(Seller seller)
    {
        sellersList.Add(seller);
        Console.WriteLine($"В магазин '{StoreName}' нанят новый продавец: {seller.Name}.");
    }

    public void ListProducts()
    {
        bool hasProducts = false;
        foreach (var seller in sellersList)
        {
            foreach (var product in seller.ProductsList)
            {
                Console.WriteLine($"- {product.GetProductInfo()}");
                hasProducts = true;
            }
        }
        if (!hasProducts)
        {
            Console.WriteLine("В магазине пока нет доступных товаров.");
        }
    }

    public string GetStoreInfo()
    {
        return $"Магазин: '{StoreName}' | Адрес: {Location} | Часы работы: {storeHours}";
    }
}
```
