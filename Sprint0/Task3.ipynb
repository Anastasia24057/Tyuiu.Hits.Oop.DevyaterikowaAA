Создать базовый класс Order в C#, который будет представлять информацию о
заказах товаров или услуг. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;

public class Item
{
    public string Name { get; set; }
    public int Amount { get; set; }
    public string Delivery { get; set; }
}

class Order
{
    public int OrderId { get; set; }
    public string CreationDate { get; set; }
    public int TotalAmount { get; set; }

    public List<string> names = new List<string>();
    public List<int> amounts = new List<int>();

    public virtual void CalculateTotal()
    {
        int sum = 0;
        for (int i = 0; i < amounts.Count; i++)
        {
            sum += amounts[i];
        }
        TotalAmount = sum;
        Console.WriteLine($"Сумма заказа: {sum}");
    }

    public virtual void AddItem(Item item)
    {
        amounts.Add(item.Amount);
        names.Add(item.Name);
        Console.WriteLine($"Добавлен товар '{item.Name}' в заказ");
    }

    public virtual void RemoveItem(Item item)
    {
        if (names.Contains(item.Name))
        {
            amounts.Remove(item.Amount);
            names.Remove(item.Name);
            Console.WriteLine($"Удален товар '{item.Name}' из заказа");
        }
        else
        {
            Console.WriteLine($"Товар '{item.Name}' не найден");
        }
    }
}

//  1. ПРОИЗВОДНЫЙ КЛАСС: ОНЛАЙН ЗАКАЗ 
class OnlineOrder : Order
{
    public string CustomerEmail { get; set; }

    
    public override void AddItem(Item item)
    {
        base.AddItem(item);
        Console.WriteLine($"Оформлена интернет-доставка в ПВЗ. Уведомление со штрихкодом отправлено на {CustomerEmail}");
    }
}

//  2. ПРОИЗВОДНЫЙ КЛАСС: ФИЗИЧЕСКИЙ ЗАКАЗ 
class PhysicalOrder : Order
{
    public string DeliveryAddress { get; set; }

    public override void RemoveItem(Item item)
    {
        base.RemoveItem(item);
        if (names.Contains(item.Name) == false) 
        {
            Console.WriteLine($"Информация о возврате: Товар будет возвращен курьером на склад с адреса {DeliveryAddress}");
        }
    }
}

//  3. ПРОИЗВОДНЫЙ КЛАСС: СПЕЦИАЛИЗИРОВАННЫЙ ЗАКАЗ
class SpecializedOrder : Order
{
    public string SpecialConditions { get; set; }
    public int DiscountPercent { get; set; } 

    public override void CalculateTotal()
    {
        int sum = 0;
        for (int i = 0; i < amounts.Count; i++)
        {
            sum += amounts[i];
        }
        
        int discount = (sum * DiscountPercent) / 100;
        TotalAmount = sum - discount;
        
        Console.WriteLine($"Сумма заказа с учетом специального условия ({SpecialConditions} -{DiscountPercent}%): {TotalAmount}");
    }
}


//  ТЕСТИРОВАНИЕ

Console.WriteLine("     Онлайн заказ");
OnlineOrder onlineOrder = new OnlineOrder { OrderId = 111, CreationDate = "09.09.26", CustomerEmail = "nastya2405@mail.com" };
Item oneItem = new Item { Name = "Помада", Amount = 450 };
Item twoItem = new Item { Name = "Тушь", Amount = 200 };

onlineOrder.AddItem(oneItem);
onlineOrder.AddItem(twoItem);
onlineOrder.CalculateTotal();

Console.WriteLine("     Физический заказ ");
PhysicalOrder physicalOrder = new PhysicalOrder { OrderId = 222, CreationDate = "09.09.26", DeliveryAddress = "ул. Ленина, д. 10" };
Item threeItem = new Item { Name = "Тональный крем", Amount = 1100 };
Item fourItem = new Item { Name = "Румяна", Amount = 600 }; 

physicalOrder.AddItem(threeItem);
physicalOrder.AddItem(fourItem);
physicalOrder.CalculateTotal();
physicalOrder.RemoveItem(threeItem); 
physicalOrder.CalculateTotal();

Console.WriteLine("     Специализированный заказ со скидкой");
SpecializedOrder specOrder = new SpecializedOrder { OrderId = 333, CreationDate = "09.09.26", SpecialConditions = "Скидка по карте лояльности", DiscountPercent = 15 };
specOrder.AddItem(fourItem); // Добавляем румяна за 600
specOrder.CalculateTotal();  // Посчитает со скидкой 15%
```
