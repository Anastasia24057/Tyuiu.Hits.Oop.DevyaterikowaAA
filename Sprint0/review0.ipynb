Разработать проект для управления растениями в саду, используя основные концепции объектно-ориентированного программирования (ООП) в C#. Создать классы, которые будут описывать разные типы растений, используя наследование и полиморфизм.

```
public class Plant
{
    public string Name { get; set; }
    public string Type { get; set; }
    public float Height { get; set; }

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Bloom()
    {
        Console.WriteLine($"Растение {Name} цветет.");
    }

    public void Grow(float growth)
    {
        Height += growth;
        Console.WriteLine($"Растение {Name} выросло на {growth} см. Теперь его высота {Height} см.");
    }
}

public class Flower : Plant
{
    public string Color { get; set; }
    public int PetalCount { get; set; }
    public string Aroma { get; set; }

    public Flower(string name, string type, float height, string color, int petalCount, string aroma) 
        : base(name, type, height)
    {
        Color = color;
        PetalCount = petalCount;
        Aroma = aroma;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Цветок {Name} полностью раскрыл бутон и приятно пахнет");
    }

    public void Wither()
    {
        Console.WriteLine($"Цветок {Name} увядает.");
    }
}

public class Tree : Plant
{
    public int Age { get; set; }
    public string LeafType { get; set; }
    public int RootDepth { get; set; }

    public Tree(string name, string type, float height, int age, string leafType, int rootDepth) 
        : base(name, type, height)
    {
        Age = age;
        LeafType = leafType;
        RootDepth = rootDepth;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Дерево {Name} зазеленело и выпустило свежие почки");
    }

    public void GrowOlder(int years)
    {
        Age += years;
        Console.WriteLine($"Дерево {Name} стало старше на {years} года. Новый возраст: {Age} лет.");
    }
}

Flower myFlower = new Flower("Лилия", "Цветок", 35f, "Фиолетовый", 5, "Нежный");
Tree myTree = new Tree("Кедр", "Дерево", 600f, 40, "Хвойные", 150);

Console.WriteLine($"Растение: {myFlower.Name}, Тип: {myFlower.Type}, Высота: {myFlower.Height} см, Цвет: {myFlower.Color}, Лепестков: {myFlower.PetalCount}, Аромат: {myFlower.Aroma}");
myFlower.Bloom();
myFlower.Grow(12f);
myFlower.Wither();

Console.WriteLine(new string('-', 100));

Console.WriteLine($"Растение: {myTree.Name}, Тип: {myTree.Type}, Высота: {myTree.Height} см, Возраст: {myTree.Age} лет, Листья: {myTree.LeafType}, Корни: {myTree.RootDepth} см");
myTree.Bloom();
myTree.Grow(45f);
myTree.GrowOlder(3);

```
