Разработать программу, описывающую взаимодействие между предприятиями, их поставщиками и сотрудниками. Основной задачей является использование интерфейсов и зависимостей между классами таким образом, чтобы их взаимодействие было максимально гибким и расширяемым.

```
using System;
using System.Collections.Generic;

// Интерфейс предприятия
public interface IEnterprise
{
    void AddSupplier(ISupplier supplier);
    void AddEmployee(IEmployee employee);
    List<ISupplier> GetSuppliers();
    List<IEmployee> GetEmployees();
    decimal CalculateTotalExpenses();
}

// Интерфейс поставщика
public interface ISupplier
{
    string GetName();
    decimal GetCost();
    string SupplierType { get; set; }
}

// Интерфейс сотрудника
public interface IEmployee
{
    string GetFullName();
    decimal GetSalary();
    string Position { get; set; }
}


public class Enterprise : IEnterprise
{
    // Списки поставщиков и сотрудников
    private List<ISupplier> suppliers = new List<ISupplier>();
    private List<IEmployee> employees = new List<IEmployee>();

    // Добавление поставщика
    public void AddSupplier(ISupplier supplier)
    {
        suppliers.Add(supplier);
    }

    // Добавление сотрудника
    public void AddEmployee(IEmployee employee)
    {
        employees.Add(employee);
    }

    // Получение списка поставщиков
    public List<ISupplier> GetSuppliers()
    {
        return suppliers;
    }

    // Получение списка сотрудников
    public List<IEmployee> GetEmployees()
    {
        return employees;
    }

    // Подсчёт общих расходов предприятия
    public decimal CalculateTotalExpenses()
    {
        decimal expenses = 0;

        foreach (ISupplier item in suppliers)
        {
            expenses += item.GetCost();
        }

        foreach (IEmployee item in employees)
        {
            expenses += item.GetSalary();
        }

        return expenses;
    }
}

// Класс поставщика
public class Supplier : ISupplier
{
    public string Name { get; set; }
    public decimal Cost { get; set; }
    public string SupplierType { get; set; }

    public Supplier(string name, decimal cost, string type)
    {
        Name = name;
        Cost = cost;
        SupplierType = type;
    }

    public string GetName()
    {
        return Name;
    }

    public decimal GetCost()
    {
        return Cost;
    }
}

// Класс сотрудника
public class Employee : IEmployee
{
    public string FullName { get; set; }
    public decimal Salary { get; set; }
    public string Position { get; set; }

    public Employee(string name, decimal salary, string position)
    {
        FullName = name;
        Salary = salary;
        Position = position;
    }

    public string GetFullName()
    {
        return FullName;
    }

    public decimal GetSalary()
    {
        return Salary;
    }
}

// Создание предприятия
IEnterprise company = new Enterprise();

// Добавление поставщиков
company.AddSupplier(new Supplier("Транспортная компания", 93000, "Перевозки"));
company.AddSupplier(new Supplier("Компания ТехноСнаб", 67000, "Компьютерная техника"));

// Добавление сотрудников
company.AddEmployee(new Employee("Девятерикова Анастасия", 95000, "Менеджер"));
company.AddEmployee(new Employee("Смирнов Алексей", 80000, "Бухгалтер"));

Console.WriteLine("Список поставщиков: ");

int count = 1;

foreach (ISupplier supplier in company.GetSuppliers())
{
    Console.WriteLine($"{count}. {supplier.GetName()} -- Стоимость услуг: {supplier.GetCost()}");
    count++;
}

Console.WriteLine();

Console.WriteLine("Список сотрудников: ");

count = 1;

foreach (IEmployee employee in company.GetEmployees())
{
    Console.WriteLine($"{count}. {employee.GetFullName()} -- Должность: {employee.Position} -- Зарплата: {employee.GetSalary()}");
    count++;
}

Console.WriteLine();

Console.WriteLine("Общие расходы: ");
Console.WriteLine(company.CalculateTotalExpenses());


```
