Cоздать классы, представляющие структуру коммерческого предприятия, используя простое, сложное и множественное наследование.

```
public interface IReportable
{
    string GenerateReport();
}

public interface ITeamLeader
{
    List<string> GetTeamMembers();
    void AddTeamMember(string member);
}

public class Employee
{
    public string Name { get; set; }
    public string Position { get; set; }
    public double Salary { get; set; }

    public Employee(string name, string position, double salary)
    {
        if (string.IsNullOrWhiteSpace(name))
            throw new ArgumentException("Имя не может быть пустым!");
        if (salary < 0)
            throw new ArgumentOutOfRangeException("Зарплата не может быть отрицательной!");

        Name = name;
        Position = position;
        Salary = salary;
    }

    public virtual void PrintInfo()
    {
        Console.WriteLine($"  Имя: {Name} | Должность: {Position} | Зарплата: {Salary:N0} руб.");
    }

    public virtual string GetRole() => "Сотрудник";
}


public class Manager : Employee
{
    public string Department { get; set; }

    public Manager(string name, string position, double salary, string department)
        : base(name, position, salary)
    {
        Department = department;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"  Отдел: {Department}");
    }

    public override string GetRole() => "Менеджер";
}

public class ProjectManager : Manager, IReportable, ITeamLeader
{
    public string ProjectName { get; set; }
    private List<string> teamMembers = new List<string>();

    public ProjectManager(string name, string position, double salary,string department, string projectName)
        : base(name, position, salary, department)
    {
        ProjectName = projectName;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"  Проект: {ProjectName} | Команда: {teamMembers.Count} чел.");
    }

    public override string GetRole() => "Проектный менеджер";

    public string GenerateReport()
        => $"  Отчёт по проекту \"{ProjectName}\": руководитель {Name}, отдел {Department}, команда {teamMembers.Count} чел.";

    public List<string> GetTeamMembers() => teamMembers;

    public void AddTeamMember(string member)
    {
        teamMembers.Add(member);
        Console.WriteLine($"  -> {member} добавлен в команду проекта \"{ProjectName}\"");
    }
}

try
{
    Console.WriteLine("==========================================================");
    Console.WriteLine("СОТРУДНИК");
    Console.WriteLine("----------------------------------------------------------");
    Employee emp = new Employee("Иван Иванов", "Разработчик", 80_000);
    emp.PrintInfo();

    Console.WriteLine("==========================================================");
    Console.WriteLine("МЕНЕДЖЕР");
    Console.WriteLine("----------------------------------------------------------");
    Manager mgr = new Manager("Пётр Петров", "Менеджер", 120_000, "IT-отдел");
    mgr.PrintInfo();

    Console.WriteLine("==========================================================");
    Console.WriteLine("ПРОЕКТНЫЙ МЕНЕДЖЕР");
    Console.WriteLine("----------------------------------------------------------");
    ProjectManager pm = new ProjectManager("Анна Сидорова", "Проектный менеджер", 150_000, "Отдел разработки", "Мобильное приложение");

    pm.AddTeamMember("Алексей");
    pm.AddTeamMember("Мария");
    pm.AddTeamMember("Дмитрий");
    pm.PrintInfo();

    Console.WriteLine("==========================================================");
    Console.WriteLine("ОТЧЁТ (IReportable)");
    Console.WriteLine("----------------------------------------------------------");
    Console.WriteLine(pm.GenerateReport());

    Console.WriteLine("==========================================================");
    Console.WriteLine("КОМАНДА (ITeamLeader)");
    Console.WriteLine("----------------------------------------------------------");
    foreach (var m in pm.GetTeamMembers())
        Console.WriteLine($"    - {m}");

    Console.WriteLine("==========================================================");
    Console.WriteLine("ЗАРПЛАТА");
    Console.WriteLine("----------------------------------------------------------");
    List<Employee> all = new List<Employee> { emp, mgr, pm };
    foreach (var e in all)
        Console.WriteLine($"  [{e.GetRole()}] {e.Name} — {e.Salary:N0} руб.");

    Console.WriteLine("==========================================================");
    Console.WriteLine("ПРОВЕРКА ВАЛИДАЦИИ");
    Console.WriteLine("----------------------------------------------------------");
    Console.WriteLine("Попытка создать сотрудника с отрицательной зарплатой...");
    Employee bad = new Employee("Тест", "Тест", -100);
}
catch (Exception ex)
{
    Console.WriteLine($"Ошибка валидации -> {ex.Message}");
}
```
